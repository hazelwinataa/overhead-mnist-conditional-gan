### 3A. Penjelasan Cara Kerja Arsitektur Generative Adversarial Network (GAN)

Generative Adversarial Network atau GAN adalah arsitektur deep learning yang digunakan untuk menghasilkan data baru yang menyerupai data asli. Pada gambar yang diberikan, GAN digunakan untuk menghasilkan gambar angka palsu (*fake image*) dari random noise. Gambar angka palsu tersebut kemudian dibandingkan dengan gambar angka asli dari *training set* melalui sebuah model pembeda yang disebut *discriminator*.

Pada arsitektur GAN, terdapat dua komponen utama, yaitu **generator** dan **discriminator**. Generator bertugas membuat data palsu, sedangkan discriminator bertugas membedakan apakah data yang diterima merupakan data asli atau data palsu. Kedua model ini dilatih secara bersamaan dengan tujuan yang saling berlawanan, sehingga proses pelatihannya disebut *adversarial training*.

Berdasarkan gambar, proses dimulai dari **random noise**. Random noise adalah kumpulan nilai acak yang digunakan sebagai input awal bagi generator. Noise ini belum memiliki bentuk gambar yang bermakna. Generator kemudian mempelajari cara mengubah noise tersebut menjadi gambar yang menyerupai data asli. Pada contoh gambar, generator menghasilkan *fake image* yang terlihat seperti gambar angka, misalnya angka 8.

Di sisi lain, terdapat **training set** yang berisi gambar-gambar asli. Pada gambar soal, training set ditunjukkan sebagai kumpulan gambar angka asli, misalnya angka 5, 6, dan 1. Gambar dari training set ini disebut **real image** karena berasal dari dataset asli. Sementara itu, gambar yang dibuat oleh generator disebut **fake image** karena bukan berasal dari dataset asli, melainkan dibuat oleh model.

Selanjutnya, discriminator menerima dua jenis input, yaitu gambar asli dari training set dan gambar palsu dari generator. Tugas discriminator adalah menentukan apakah gambar yang masuk termasuk **real** atau **fake**. Dengan kata lain, discriminator bekerja seperti model binary classification karena hanya memiliki dua kemungkinan output, yaitu real atau fake.

Dalam proses pelatihan discriminator, gambar asli dari training set diberi target **real = 1**, sedangkan gambar palsu dari generator diberi target **fake = 0**. Discriminator akan belajar agar output-nya mendekati 1 ketika menerima gambar asli dan mendekati 0 ketika menerima gambar palsu. Jika discriminator dapat membedakan real image dan fake image dengan benar, maka discriminator semakin baik.

Namun, tujuan generator berbeda dari discriminator. Generator ingin menghasilkan fake image yang terlihat semakin realistis sehingga discriminator sulit membedakannya dari gambar asli. Oleh karena itu, ketika generator dilatih, fake image yang dihasilkannya dibuat seolah-olah memiliki target **real = 1**. Artinya, generator dilatih agar gambar palsu yang dibuatnya dapat menipu discriminator dan diklasifikasikan sebagai gambar asli.

Proses training GAN dilakukan secara bergantian. Pertama, generator menghasilkan fake image dari random noise. Kedua, discriminator dilatih menggunakan gabungan real image dari training set dan fake image dari generator. Pada tahap ini, discriminator belajar membedakan gambar asli dan gambar palsu. Ketiga, generator dilatih agar menghasilkan fake image yang lebih baik, dengan tujuan membuat discriminator menganggap fake image tersebut sebagai real image.

Hubungan antara generator dan discriminator dapat dianalogikan seperti pemalsu dan pemeriksa. Generator berperan sebagai pemalsu yang berusaha membuat gambar palsu semirip mungkin dengan gambar asli. Discriminator berperan sebagai pemeriksa yang berusaha mengenali apakah gambar tersebut asli atau palsu. Jika discriminator berhasil membedakan fake image, maka generator perlu memperbaiki hasilnya. Sebaliknya, jika generator berhasil menipu discriminator, maka generator dianggap semakin baik.

Karena kedua model saling berkompetisi, nilai loss generator dan discriminator dapat saling memengaruhi. Jika discriminator terlalu mudah membedakan gambar asli dan palsu, maka generator loss dapat menjadi tinggi karena generator gagal menipu discriminator. Sebaliknya, jika generator semakin mampu menghasilkan gambar realistis, discriminator akan semakin sulit membedakan real image dan fake image.

Tujuan akhir dari GAN adalah menghasilkan generator yang mampu membuat gambar baru dengan distribusi yang mirip dengan data asli. Dalam contoh gambar soal, generator diharapkan mampu menghasilkan gambar angka palsu yang semakin lama semakin menyerupai gambar angka asli dari training set. Jika proses training berjalan dengan baik, fake image yang awalnya hanya terlihat seperti noise akan berkembang menjadi gambar yang lebih jelas dan realistis.

Secara sederhana, alur kerja GAN pada gambar dapat diringkas sebagai berikut:

1. Random noise diberikan sebagai input ke generator.
2. Generator mengubah random noise menjadi fake image.
3. Training set menyediakan real image sebagai pembanding.
4. Discriminator menerima real image dari training set dan fake image dari generator.
5. Discriminator memprediksi apakah gambar tersebut real atau fake.
6. Discriminator dilatih agar semakin baik membedakan real image dan fake image.
7. Generator dilatih agar fake image yang dibuat semakin realistis dan mampu menipu discriminator.
8. Proses ini diulang berkali-kali sampai generator dapat menghasilkan gambar yang menyerupai data asli.

Secara matematis, tujuan GAN dapat dijelaskan sebagai permainan minimax antara generator dan discriminator. Discriminator berusaha memaksimalkan kemampuannya dalam membedakan data asli dan data palsu, sedangkan generator berusaha meminimalkan kemampuan discriminator dalam mengenali gambar palsu. Objective GAN secara umum dapat ditulis sebagai:

```
min G max D V(D, G) =
E_x~pdata(x) [log D(x)] + E_z~pz(z) [log(1 - D(G(z)))]
```

Pada persamaan tersebut, `x` adalah data asli dari training set, `z` adalah random noise, `G(z)` adalah fake image yang dihasilkan generator, dan `D(x)` adalah probabilitas bahwa suatu gambar dianggap real oleh discriminator. Discriminator ingin memberikan probabilitas tinggi untuk gambar asli dan probabilitas rendah untuk gambar palsu. Sebaliknya, generator ingin membuat `D(G(z))` bernilai tinggi agar fake image dianggap sebagai real image.

Dengan demikian, gambar pada soal menunjukkan cara kerja dasar GAN untuk image generation. Generator belajar mengubah random noise menjadi fake image, sedangkan discriminator belajar membedakan gambar asli dari training set dan gambar palsu dari generator. Melalui proses adversarial training yang dilakukan berulang, generator diharapkan mampu menghasilkan gambar baru yang semakin realistis dan semakin mirip dengan data asli.

#### Simulasi Contoh Cara Kerja GAN

Sebagai contoh sederhana, misalkan dataset training berisi gambar angka asli seperti angka 1, 5, 6, dan 8. Pada awal training, generator belum mengetahui seperti apa bentuk angka yang benar. Generator hanya menerima random noise, yaitu kumpulan angka acak, kemudian mencoba mengubahnya menjadi gambar.

Misalnya pada iterasi awal:

```
Random noise → Generator → Fake image yang masih terlihat seperti noise
```

Kemudian discriminator menerima dua jenis gambar:

```
Real image dari training set  → label real = 1
Fake image dari generator     → label fake = 0
```

Contoh simulasi input ke discriminator dapat dijelaskan sebagai berikut:

- Gambar angka 5 asli berasal dari training set, sehingga target yang benar adalah real = 1.
- Gambar angka 1 asli juga berasal dari training set, sehingga target yang benar adalah real = 1.
- Gambar angka palsu yang masih mirip noise berasal dari generator, sehingga target yang benar adalah fake = 0.
- Gambar angka palsu yang belum jelas juga berasal dari generator, sehingga target yang benar adalah fake = 0.

Pada tahap awal, discriminator biasanya mudah membedakan gambar asli dan gambar palsu karena fake image dari generator masih belum jelas. Misalnya discriminator menghasilkan prediksi seperti berikut:

- Untuk angka 5 asli, target benarnya adalah 1 dan prediksi discriminator adalah 0.95. Artinya, discriminator benar karena gambar tersebut dianggap real.
- Untuk angka 1 asli, target benarnya adalah 1 dan prediksi discriminator adalah 0.92. Artinya, discriminator benar karena gambar tersebut dianggap real.
- Untuk fake image awal, target benarnya adalah 0 dan prediksi discriminator adalah 0.08. Artinya, discriminator benar karena gambar tersebut dianggap fake.
- Untuk fake image awal lainnya, target benarnya adalah 0 dan prediksi discriminator adalah 0.12. Artinya, discriminator juga benar karena gambar tersebut dianggap fake.

Pada kondisi ini, discriminator sudah cukup baik karena mampu mengenali gambar asli sebagai real dan gambar palsu sebagai fake. Namun, generator masih buruk karena fake image yang dibuatnya masih mudah dikenali sebagai palsu.

Setelah discriminator dilatih, generator juga dilatih. Saat melatih generator, fake image yang dibuat generator diberi target **real = 1**. Hal ini dilakukan karena tujuan generator adalah membuat gambar palsu yang dapat menipu discriminator.

Contohnya:

```
Random noise → Generator → Fake image
Fake image → Discriminator → diharapkan dianggap real
```

Pada tahap ini, target untuk fake image bukan lagi 0, tetapi dibuat menjadi 1.

Contoh dampaknya terhadap generator dapat dijelaskan sebagai berikut:

- Jika gambar dari generator memiliki target 1, tetapi prediksi discriminator hanya 0.10, maka loss generator besar.
- Jika gambar dari generator memiliki target 1, tetapi prediksi discriminator hanya 0.25, maka generator masih perlu belajar.
- Jika gambar dari generator memiliki target 1 dan prediksi discriminator sudah mencapai 0.70, maka generator mulai membaik.
- Jika gambar dari generator memiliki target 1 dan prediksi discriminator mencapai 0.90, maka generator berhasil menipu discriminator.

Jika discriminator memberi nilai 0.10 pada fake image, artinya discriminator yakin gambar tersebut palsu. Karena target generator adalah 1, maka loss generator menjadi besar. Generator kemudian memperbarui bobotnya agar pada iterasi berikutnya dapat menghasilkan gambar yang lebih realistis.

Setelah banyak epoch, fake image yang awalnya hanya terlihat seperti noise mulai membentuk pola yang lebih jelas. Misalnya:

```
Epoch awal:
Random noise → gambar acak tidak jelas

Epoch menengah:
Random noise → mulai muncul garis seperti angka

Epoch akhir:
Random noise → gambar terlihat seperti angka 8 atau angka lain yang menyerupai training set
```

Simulasi perkembangan hasil generator dapat dijelaskan sebagai berikut:

- Pada awal training, hasil generator masih berupa gambar seperti noise. Pada tahap ini, discriminator masih mudah membedakan real image dan fake image.
- Pada tengah training, hasil generator mulai memunculkan pola seperti angka. Pada tahap ini, discriminator mulai lebih sulit membedakan gambar asli dan gambar palsu.
- Pada akhir training, fake image yang dihasilkan generator semakin mirip dengan angka asli. Pada tahap ini, discriminator semakin sulit menentukan apakah gambar tersebut real atau fake.

Dengan demikian, proses GAN dapat dipahami sebagai latihan berulang antara generator dan discriminator. Discriminator belajar membedakan gambar asli dan palsu, sedangkan generator belajar menghasilkan gambar palsu yang semakin realistis. Jika proses training berjalan baik, generator akhirnya mampu menghasilkan gambar baru yang menyerupai gambar asli dari training set.

Berdasarkan gambar arsitektur pada soal 3B, model yang digunakan lebih tepat disebut **Conditional GAN (CGAN)**, bukan vanilla GAN biasa. Hal ini terlihat dari adanya bagian **Embedding(labels)** pada generator dan discriminator. Pada vanilla GAN, generator hanya menerima random noise sebagai input dan tidak memiliki informasi mengenai kelas gambar yang harus dibuat. Akibatnya, generator hanya belajar menghasilkan gambar dari distribusi umum dataset tanpa kontrol kelas tertentu.

Sebaliknya, pada Conditional GAN, generator menerima dua input, yaitu **random noise** dan **class label**. Random noise digunakan sebagai sumber variasi gambar, sedangkan class label digunakan sebagai kondisi agar generator dapat menghasilkan gambar sesuai kelas yang diminta. Dengan demikian, model dapat diarahkan untuk membuat gambar Parking Lot atau Storage Tanks secara lebih terkontrol.

Discriminator pada Conditional GAN juga menerima dua input, yaitu **image** dan **class label**. Artinya, discriminator tidak hanya belajar membedakan apakah gambar tersebut real atau fake, tetapi juga mempertimbangkan label kelas yang diberikan. Real image dari dataset diberi target real, sedangkan fake image dari generator diberi target fake. Generator kemudian dilatih agar fake image yang dihasilkannya dapat dianggap real oleh discriminator.

Penggunaan CGAN sesuai dengan karakter data yang digunakan karena dataset memiliki dua kelas berbeda dengan pola visual yang berbeda. Parking Lot umumnya memiliki pola garis, area parkir, dan tekstur yang berulang, sedangkan Storage Tanks memiliki ciri visual berupa objek berbentuk tangki atau lingkaran dari tampak atas. Jika hanya menggunakan vanilla GAN, model tidak memiliki kontrol untuk menentukan kelas gambar yang dihasilkan. Dengan menggunakan CGAN, label kelas dapat membantu generator mempelajari karakter visual masing-masing kelas secara lebih terarah.

Pada bagian 3B, arsitektur baseline dibuat mengikuti gambar soal, yaitu menggunakan layer Linear atau Dense pada generator dan discriminator. Generator menerima random noise dan label, lalu menghasilkan fake image berukuran 28 × 28 × 1. Discriminator menerima image dan label, lalu menghasilkan output berupa probabilitas apakah gambar tersebut real atau fake.

Evaluasi model dilakukan menggunakan **Fréchet Inception Distance (FID)** pada data test. FID digunakan untuk mengukur seberapa mirip distribusi gambar hasil generator dengan distribusi gambar asli. Semakin kecil nilai FID, semakin baik kualitas dan keragaman gambar yang dihasilkan. Oleh karena itu, nilai FID dari baseline CGAN pada bagian 3B akan digunakan sebagai dasar pembanding untuk model yang dimodifikasi pada bagian 3C.

## Load IDX Dataset

````python
import struct
import numpy as np
from pathlib import Path

DATA_DIR = Path("data_soal_2")

TRAIN_IMAGES_PATH = DATA_DIR / "train-images-idx3-ubyte"
TRAIN_LABELS_PATH = DATA_DIR / "train-labels-idx1-ubyte"
TEST_IMAGES_PATH = DATA_DIR / "t10k-images-idx3-ubyte"
TEST_LABELS_PATH = DATA_DIR / "t10k-labels-idx1-ubyte"

print("File exists check:")
print("Train images:", TRAIN_IMAGES_PATH.exists())
print("Train labels:", TRAIN_LABELS_PATH.exists())
print("Test images :", TEST_IMAGES_PATH.exists())
print("Test labels :", TEST_LABELS_PATH.exists())


def load_idx_images(file_path):
    with open(file_path, "rb") as f:
        magic, num_images, rows, cols = struct.unpack(">IIII", f.read(16))
        images = np.frombuffer(f.read(), dtype=np.uint8)
        images = images.reshape(num_images, rows, cols)
    return images


def load_idx_labels(file_path):
    with open(file_path, "rb") as f:
        magic, num_labels = struct.unpack(">II", f.read(8))
        labels = np.frombuffer(f.read(), dtype=np.uint8)
    return labels


X_train_full = load_idx_images(TRAIN_IMAGES_PATH)
y_train_full = load_idx_labels(TRAIN_LABELS_PATH)

X_test_full = load_idx_images(TEST_IMAGES_PATH)
y_test_full = load_idx_labels(TEST_LABELS_PATH)

print("\nTrain images shape:", X_train_full.shape)
print("Train labels shape:", y_train_full.shape)
print("Test images shape :", X_test_full.shape)
print("Test labels shape :", y_test_full.shape)
````

## Filter Kelas Parking Lot dan Storage Tanks

````python
SELECTED_CLASSES = {
    4: "Parking Lot",
    9: "Storage Tanks"
}

NEW_LABEL_MAP = {
    4: 0,
    9: 1
}

CLASS_NAMES = {
    0: "Parking Lot",
    1: "Storage Tanks"
}


def filter_selected_classes(images, labels):
    mask = np.isin(labels, list(SELECTED_CLASSES.keys()))
    selected_images = images[mask]
    selected_labels_original = labels[mask]

    selected_labels = np.array(
        [NEW_LABEL_MAP[label] for label in selected_labels_original],
        dtype=np.int32
    )

    return selected_images, selected_labels


X_train, y_train = filter_selected_classes(X_train_full, y_train_full)
X_test, y_test = filter_selected_classes(X_test_full, y_test_full)

print("Selected train shape:", X_train.shape)
print("Selected test shape :", X_test.shape)

print("\nTrain class distribution:")
for label in sorted(np.unique(y_train)):
    print(f"{label} - {CLASS_NAMES[label]}: {np.sum(y_train == label)}")

print("\nTest class distribution:")
for label in sorted(np.unique(y_test)):
    print(f"{label} - {CLASS_NAMES[label]}: {np.sum(y_test == label)}")
````

## Build Generator

````python
NOISE_DIM = 100


def build_generator(noise_dim=NOISE_DIM, num_classes=NUM_CLASSES):
    noise_input = Input(shape=(noise_dim,), name="noise_input")
    label_input = Input(shape=(1,), dtype="int32", name="label_input")

    label_embedding = Embedding(num_classes, num_classes)(label_input)
    label_embedding = Flatten()(label_embedding)

    x = Concatenate()([noise_input, label_embedding])

    x = Dense(128)(x)
    x = LeakyReLU(alpha=0.2)(x)

    x = Dense(256)(x)
    x = LeakyReLU(alpha=0.2)(x)

    x = Dense(512)(x)
    x = LeakyReLU(alpha=0.2)(x)

    x = Dense(1024)(x)
    x = LeakyReLU(alpha=0.2)(x)

    x = Dense(IMG_SIZE * IMG_SIZE * CHANNELS, activation="tanh")(x)
    output_image = Reshape((IMG_SIZE, IMG_SIZE, CHANNELS))(x)

    generator = Model([noise_input, label_input], output_image, name="Baseline_Generator")
    return generator


generator = build_generator()
generator.summary()
````

Pada bagian ini, saya membangun baseline generator untuk Conditional GAN. Generator menerima dua input, yaitu random noise dan label kelas. Random noise berfungsi sebagai sumber variasi agar model dapat menghasilkan gambar yang berbeda-beda, sedangkan label kelas digunakan agar generator dapat diarahkan untuk menghasilkan gambar sesuai kelas tertentu, yaitu Parking Lot atau Storage Tanks.

Label kelas terlebih dahulu diproses menggunakan Embedding layer. Tujuannya adalah mengubah label numerik menjadi representasi vektor yang dapat dipelajari oleh model. Setelah itu, hasil embedding label digabungkan dengan random noise menggunakan Concatenate. Dengan cara ini, generator tidak hanya menghasilkan gambar secara acak, tetapi juga mempertimbangkan informasi kelas yang diberikan.

Setelah noise dan label digabungkan, data dilewatkan ke beberapa Dense layer dengan ukuran 128, 256, 512, dan 1024. Susunan ini mengikuti arsitektur pada gambar soal, yaitu generator berbasis Linear/Dense layer. Semakin dalam layer, semakin besar kapasitas model untuk mempelajari pola dari data gambar.

Pada setiap Dense layer, saya menambahkan aktivasi LeakyReLU. Walaupun pada gambar soal tidak dituliskan secara eksplisit, aktivasi tetap diperlukan dalam implementasi neural network. Jika hanya menggunakan Dense layer tanpa aktivasi, maka rangkaian layer tersebut pada dasarnya hanya menjadi transformasi linear. Dengan adanya aktivasi LeakyReLU, model dapat mempelajari pola non-linear yang lebih kompleks dari gambar. LeakyReLU juga dipilih karena umum digunakan pada GAN dan dapat membantu mengurangi masalah neuron mati dibandingkan ReLU biasa.

Pada layer terakhir, generator menggunakan Dense dengan jumlah neuron sebesar 28 × 28 × 1, karena ukuran citra Overhead-MNIST adalah 28×28 grayscale. Aktivasi yang digunakan pada output adalah tanh karena data gambar sudah dinormalisasi ke rentang -1 sampai 1. Setelah itu, output diubah kembali menggunakan Reshape menjadi bentuk citra 28×28×1.

Baseline generator ini tidak menggunakan Dropout karena bagian baseline dibuat agar tetap mendekati arsitektur pada gambar soal. Teknik tambahan seperti Dropout lebih tepat digunakan pada modified model, karena bagian modified memang bertujuan untuk melakukan perubahan arsitektur dan tuning hyperparameter.

## Build Discriminator

````python
def build_discriminator(num_classes=NUM_CLASSES):
    image_input = Input(shape=(IMG_SIZE, IMG_SIZE, CHANNELS), name="image_input")
    label_input = Input(shape=(1,), dtype="int32", name="label_input")

    image_flat = Flatten()(image_input)

    label_embedding = Embedding(num_classes, num_classes)(label_input)
    label_embedding = Flatten()(label_embedding)

    x = Concatenate()([image_flat, label_embedding])

    x = Dense(512)(x)
    x = LeakyReLU(alpha=0.2)(x)

    x = Dense(1024)(x)
    x = LeakyReLU(alpha=0.2)(x)

    x = Dense(1024)(x)
    x = LeakyReLU(alpha=0.2)(x)

    x = Dense(512)(x)
    x = LeakyReLU(alpha=0.2)(x)

    output = Dense(1, activation="sigmoid")(x)

    discriminator = Model(
        [image_input, label_input],
        output,
        name="Baseline_Discriminator"
    )

    return discriminator


discriminator = build_discriminator()

discriminator.compile(
    loss="binary_crossentropy",
    optimizer=Adam(learning_rate=0.00005, beta_1=0.5),
    metrics=["accuracy"]
)

discriminator.summary()
````

Pada bagian ini, saya membangun baseline discriminator untuk Conditional GAN. Discriminator menerima dua input, yaitu gambar dan label kelas. Gambar yang masuk dapat berupa real image dari dataset maupun fake image hasil generator, sedangkan label digunakan sebagai kondisi agar discriminator dapat menilai apakah gambar tersebut sesuai dengan kelas yang diberikan.

Input gambar berukuran 28×28×1 terlebih dahulu diubah menjadi bentuk vektor menggunakan Flatten. Karena gambar Overhead-MNIST berbentuk grayscale 28×28, maka hasil flatten menjadi 784 nilai piksel. Selain itu, label kelas juga diproses menggunakan Embedding layer agar label numerik dapat diubah menjadi representasi vektor yang dapat dipelajari oleh model.

Setelah gambar dan label diproses, keduanya digabungkan menggunakan Concatenate. Dengan cara ini, discriminator tidak hanya menilai apakah gambar terlihat real atau fake, tetapi juga mempertimbangkan informasi label kelas yang diberikan. Hal ini sesuai dengan konsep Conditional GAN, karena discriminator menerima image dan condition label sebagai input.

Setelah proses penggabungan, data dilewatkan ke beberapa Dense layer dengan ukuran 512, 1024, 1024, dan 512. Susunan ini mengikuti arsitektur discriminator pada gambar soal yang menggunakan Linear/Dense layer. Pada setiap Dense layer, saya menambahkan aktivasi LeakyReLU. Walaupun aktivasi tidak ditulis secara eksplisit pada gambar soal, aktivasi tetap diperlukan dalam implementasi neural network agar model dapat mempelajari pola non-linear. LeakyReLU dipilih karena umum digunakan pada GAN dan dapat membantu menjaga aliran gradien selama training.

Layer terakhir menggunakan Dense dengan 1 neuron dan aktivasi sigmoid. Output ini merepresentasikan probabilitas apakah gambar yang diberikan adalah real atau fake. Nilai mendekati 1 menunjukkan bahwa discriminator menganggap gambar tersebut real, sedangkan nilai mendekati 0 menunjukkan bahwa gambar dianggap fake.

Pada baseline discriminator ini, Dropout tidak digunakan agar struktur model tetap mendekati arsitektur pada gambar soal. Teknik tambahan seperti Dropout lebih tepat digunakan pada modified model karena bagian modified memang bertujuan untuk melakukan perubahan arsitektur dan tuning hyperparameter.

Model discriminator kemudian dikompilasi menggunakan binary crossentropy karena tugas discriminator adalah binary classification, yaitu membedakan real image dan fake image. Optimizer yang digunakan adalah Adam dengan learning rate 0.00005 dan beta\_1 0.5. Nilai beta\_1 0.5 umum digunakan pada training GAN karena dapat membantu proses adversarial training menjadi lebih stabil. Metric accuracy hanya digunakan untuk memantau performa discriminator selama training, sedangkan evaluasi utama kualitas gambar tetap menggunakan FID pada data test.

## Build Combined CGAN

````python
def build_cgan(generator, discriminator, noise_dim=NOISE_DIM):
    discriminator.trainable = False

    noise_input = Input(shape=(noise_dim,), name="cgan_noise_input")
    label_input = Input(shape=(1,), dtype="int32", name="cgan_label_input")

    generated_image = generator([noise_input, label_input])
    validity = discriminator([generated_image, label_input])

    cgan = Model([noise_input, label_input], validity, name="Baseline_CGAN")

    cgan.compile(
        loss="binary_crossentropy",
        optimizer=Adam(learning_rate=0.00005, beta_1=0.5)
    )

    return cgan


cgan = build_cgan(generator, discriminator)
cgan.summary()
````

Pada tahap ini, generator dan discriminator digabung menjadi satu model bernama **Baseline\_CGAN**. Model gabungan ini digunakan khusus untuk melatih generator agar mampu menghasilkan fake image yang dianggap real oleh discriminator.

Pertama, `discriminator.trainable = False` digunakan untuk membekukan bobot discriminator saat model CGAN dilatih. Hal ini penting karena pada tahap training generator, bobot yang ingin diperbarui hanya bobot generator. Discriminator hanya digunakan sebagai penilai apakah gambar yang dibuat generator terlihat real atau fake, tetapi bobot discriminator tidak ikut berubah pada tahap ini.

Model CGAN menerima dua input, yaitu random noise dan class label. Random noise digunakan sebagai sumber variasi gambar, sedangkan class label digunakan agar generator dapat menghasilkan gambar sesuai kelas tertentu, yaitu Parking Lot atau Storage Tanks.

Alur pada combined CGAN adalah sebagai berikut:

```
noise + label → generator → generated image
generated image + label → discriminator → validity
```

Output dari generator berupa `generated_image`, yaitu fake image yang dibuat berdasarkan noise dan label. Fake image tersebut kemudian diberikan ke discriminator bersama label yang sama. Discriminator menghasilkan output `validity`, yaitu probabilitas apakah gambar tersebut dianggap real atau fake.

Model gabungan ini kemudian dikompilasi menggunakan loss `binary_crossentropy`, karena target training-nya tetap berupa klasifikasi biner real atau fake. Optimizer yang digunakan adalah Adam dengan learning rate 0.00005 dan `beta_1=0.5` agar proses training GAN lebih stabil.

Pada saat training CGAN, target untuk fake image dibuat menjadi **real = 1**. Tujuannya adalah agar generator belajar menghasilkan fake image yang mampu menipu discriminator. Jika discriminator menganggap fake image sebagai fake, maka generator mendapatkan loss yang besar. Sebaliknya, jika discriminator menganggap fake image sebagai real, maka generator dianggap berhasil.

Dengan demikian, combined CGAN ini bukan digunakan untuk melatih discriminator, tetapi digunakan untuk melatih generator melalui feedback dari discriminator. Struktur ini penting dalam GAN karena generator tidak dibandingkan langsung dengan gambar asli secara pixel-by-pixel, melainkan belajar dari penilaian discriminator.

## Function Visualisasi Generated Images

````python
def show_generated_images(generator, epoch, noise_dim=NOISE_DIM, examples_per_class=5):
    plt.figure(figsize=(10, 4))

    plot_index = 1

    for class_label in range(NUM_CLASSES):
        noise = np.random.normal(0, 1, (examples_per_class, noise_dim))
        labels = np.full((examples_per_class, 1), class_label)

        generated_images = generator.predict([noise, labels], verbose=0)
        generated_images = (generated_images + 1) / 2

        for i in range(examples_per_class):
            plt.subplot(NUM_CLASSES, examples_per_class, plot_index)
            plt.imshow(generated_images[i].squeeze(), cmap="gray")
            plt.title(CLASS_NAMES[class_label])
            plt.axis("off")
            plot_index += 1

    plt.suptitle(f"Generated Images at Epoch {epoch}")
    plt.tight_layout()
    plt.show()
````

## Custom Training Function

````python
def train_cgan(
    generator,
    discriminator,
    cgan,
    X_train,
    y_train,
    epochs=100,
    batch_size=128,
    steps_per_epoch=100,
    noise_dim=NOISE_DIM,
    display_interval=10
):
    half_batch = batch_size // 2

    d_losses = []
    d_accuracies = []
    g_losses = []

    for epoch in range(1, epochs + 1):
        epoch_d_loss = []
        epoch_d_acc = []
        epoch_g_loss = []

        for step in range(steps_per_epoch):
            # Train discriminator with real images
            real_idx = np.random.randint(0, X_train.shape[0], half_batch)
            real_images = X_train[real_idx]
            real_labels = y_train[real_idx].reshape(-1, 1)
            real_targets = np.ones((half_batch, 1))

            # Train discriminator with fake images
            noise = np.random.normal(0, 1, (half_batch, noise_dim))
            fake_labels = np.random.randint(0, NUM_CLASSES, half_batch).reshape(-1, 1)
            fake_images = generator.predict([noise, fake_labels], verbose=0)
            fake_targets = np.zeros((half_batch, 1))

            discriminator.trainable = True
            d_loss_real = discriminator.train_on_batch([real_images, real_labels], real_targets)
            d_loss_fake = discriminator.train_on_batch([fake_images, fake_labels], fake_targets)

            d_loss = 0.5 * np.add(d_loss_real, d_loss_fake)

            # Train generator through combined CGAN
            noise = np.random.normal(0, 1, (batch_size, noise_dim))
            sampled_labels = np.random.randint(0, NUM_CLASSES, batch_size).reshape(-1, 1)
            valid_targets = np.ones((batch_size, 1))

            discriminator.trainable = False
            g_loss = cgan.train_on_batch([noise, sampled_labels], valid_targets)

            epoch_d_loss.append(d_loss[0])
            epoch_d_acc.append(d_loss[1])
            epoch_g_loss.append(g_loss)

        d_losses.append(np.mean(epoch_d_loss))
        d_accuracies.append(np.mean(epoch_d_acc))
        g_losses.append(np.mean(epoch_g_loss))

        if epoch % display_interval == 0 or epoch == 1:
            print(
                f"Epoch {epoch}/{epochs} | "
                f"D Loss: {d_losses[-1]:.4f} | "
                f"D Acc: {d_accuracies[-1]:.4f} | "
                f"G Loss: {g_losses[-1]:.4f}"
            )
            show_generated_images(generator, epoch, noise_dim=noise_dim)

    return d_losses, d_accuracies, g_losses
````

Pada tahap ini, dibuat custom training function untuk melatih baseline Conditional GAN. Training GAN tidak cukup dilakukan menggunakan `model.fit()` biasa karena terdapat dua model yang harus dilatih secara bergantian, yaitu discriminator dan generator. Discriminator dilatih untuk membedakan real image dan fake image, sedangkan generator dilatih agar fake image yang dibuatnya dapat dianggap real oleh discriminator.

Fungsi `train_cgan` menerima beberapa input utama, yaitu generator, discriminator, model gabungan CGAN, data training, label training, jumlah epoch, batch size, jumlah step per epoch, dimensi noise, dan interval untuk menampilkan hasil gambar. Parameter `half_batch` digunakan karena dalam satu step training discriminator, data dibagi menjadi dua bagian: setengah batch untuk real image dan setengah batch untuk fake image.

Pada setiap epoch, fungsi ini menyimpan tiga nilai utama, yaitu discriminator loss, discriminator accuracy, dan generator loss. Nilai-nilai tersebut disimpan agar proses training dapat dianalisis setelah training selesai.

Langkah pertama dalam setiap step adalah melatih discriminator menggunakan real image. Kode mengambil beberapa gambar asli secara acak dari `X_train`, lalu mengambil label kelas yang sesuai dari `y_train`. Gambar asli ini diberi target `real = 1`, karena gambar tersebut berasal dari dataset asli.

Setelah itu, generator digunakan untuk membuat fake image. Random noise dibuat menggunakan distribusi normal, lalu label kelas juga dibuat secara acak dari dua kelas yang tersedia, yaitu Parking Lot dan Storage Tanks. Noise dan label tersebut diberikan ke generator untuk menghasilkan fake image. Karena gambar ini merupakan hasil buatan generator, target untuk discriminator adalah `fake = 0`.

Discriminator kemudian dilatih dua kali dalam satu step, yaitu sekali menggunakan real image dan sekali menggunakan fake image. Loss dari real image dan fake image kemudian dirata-ratakan untuk mendapatkan nilai discriminator loss pada step tersebut. Hal ini dilakukan agar performa discriminator mencerminkan kemampuannya dalam mengenali kedua jenis gambar, yaitu real dan fake.

Setelah discriminator dilatih, tahap berikutnya adalah melatih generator melalui model gabungan CGAN. Pada tahap ini, random noise dan label baru dibuat kembali. Generator menghasilkan fake image berdasarkan noise dan label tersebut, kemudian fake image dinilai oleh discriminator melalui model CGAN.

Saat melatih generator, target untuk fake image dibuat menjadi `real = 1`. Hal ini dilakukan karena tujuan generator adalah menipu discriminator agar fake image dianggap sebagai real image. Pada tahap ini, discriminator dibuat tidak trainable agar bobot yang diperbarui hanya bobot generator. Dengan demikian, generator belajar memperbaiki kualitas gambar berdasarkan feedback dari discriminator.

Setelah seluruh step dalam satu epoch selesai, nilai rata-rata discriminator loss, discriminator accuracy, dan generator loss disimpan. Jika epoch mencapai interval tertentu, fungsi akan menampilkan informasi loss dan accuracy, lalu memanggil fungsi `show_generated_images` untuk melihat perkembangan gambar yang dihasilkan oleh generator.

Dengan demikian, fungsi ini menjalankan proses training CGAN secara lengkap. Discriminator belajar membedakan gambar asli dan gambar palsu, sedangkan generator belajar menghasilkan fake image yang semakin realistis dan sesuai dengan label kelas yang diberikan.

## Train Baseline CGAN

````python
EPOCHS = 100
BATCH_SIZE = 128
STEPS_PER_EPOCH = 100

d_losses, d_accuracies, g_losses = train_cgan(
    generator=generator,
    discriminator=discriminator,
    cgan=cgan,
    X_train=X_train,
    y_train=y_train,
    epochs=EPOCHS,
    batch_size=BATCH_SIZE,
    steps_per_epoch=STEPS_PER_EPOCH,
    noise_dim=NOISE_DIM,
    display_interval=10
)
````

## Generate Fake Images untuk Data Test

````python
def generate_images_by_labels(generator, labels, noise_dim=NOISE_DIM, batch_size=256):
    generated_images = []

    labels = labels.reshape(-1, 1)

    for start in range(0, len(labels), batch_size):
        end = start + batch_size
        current_labels = labels[start:end]

        noise = np.random.normal(0, 1, (len(current_labels), noise_dim))
        fake_batch = generator.predict([noise, current_labels], verbose=0)

        generated_images.append(fake_batch)

    generated_images = np.concatenate(generated_images, axis=0)
    return generated_images


X_fake_test = generate_images_by_labels(generator, y_test, noise_dim=NOISE_DIM)

print("Real test shape:", X_test.shape)
print("Fake test shape:", X_fake_test.shape)
````

Pada tahap ini, generator digunakan untuk membuat **fake images** berdasarkan label pada data test. Fungsi `generate_images_by_labels` menerima generator, label test, dimensi noise, dan batch size.

Pertama, label test diubah menjadi bentuk `(jumlah_data, 1)` agar sesuai dengan format input label pada CGAN. Setelah itu, proses generate gambar dilakukan per batch supaya lebih efisien dan tidak membebani memori.

Untuk setiap batch label, random noise dibuat menggunakan distribusi normal. Noise tersebut kemudian digabungkan dengan label dan diberikan ke generator. Dengan cara ini, generator menghasilkan fake image sesuai label kelas yang diberikan, yaitu **Parking Lot** atau **Storage Tanks**.

Fake image dari setiap batch disimpan ke dalam list, lalu digabungkan kembali menggunakan `np.concatenate`. Hasil akhirnya adalah `X_fake_test`, yaitu kumpulan gambar palsu yang jumlahnya sama dengan data test.

Output menunjukkan bahwa bentuk data real test adalah:

```
(1784, 28, 28, 1)
```

dan bentuk data fake test juga:

## FID Function

````python
def calculate_fid(real_images, fake_images, batch_size=128, eps=1e-6):
    inception_model = InceptionV3(
        include_top=False,
        pooling="avg",
        input_shape=(75, 75, 3),
        weights="imagenet"
    )

    real_images_prepared = prepare_images_for_inception(real_images)
    fake_images_prepared = prepare_images_for_inception(fake_images)

    real_features = inception_model.predict(real_images_prepared, batch_size=batch_size, verbose=1)
    fake_features = inception_model.predict(fake_images_prepared, batch_size=batch_size, verbose=1)

    mu_real = np.mean(real_features, axis=0)
    mu_fake = np.mean(fake_features, axis=0)

    sigma_real = np.cov(real_features, rowvar=False)
    sigma_fake = np.cov(fake_features, rowvar=False)

    # Add small epsilon to diagonal for numerical stability
    sigma_real += np.eye(sigma_real.shape[0]) * eps
    sigma_fake += np.eye(sigma_fake.shape[0]) * eps

    mean_diff = mu_real - mu_fake
    covmean = sqrtm(sigma_real.dot(sigma_fake))

    if np.iscomplexobj(covmean):
        covmean = covmean.real

    fid = mean_diff.dot(mean_diff) + np.trace(sigma_real + sigma_fake - 2 * covmean)

    return float(fid)
````

## Calculate Baseline FID

````python
baseline_fid = calculate_fid(X_test, X_fake_test)

print(f"Baseline CGAN FID on test data: {baseline_fid:.4f}")
````

### Build Modified Generator

````python
def build_modified_generator(
    noise_dim,
    num_classes=NUM_CLASSES,
    name="Modified_Generator"
):
    noise_input = Input(shape=(noise_dim,), name="noise_input")
    label_input = Input(shape=(1,), dtype="int32", name="label_input")

    label_embedding = Embedding(num_classes, 7 * 7)(label_input)
    label_embedding = Flatten()(label_embedding)
    label_map = Reshape((7, 7, 1))(label_embedding)

    x = Dense(7 * 7 * 128, use_bias=False)(noise_input)
    x = BatchNormalization()(x)
    x = LeakyReLU(negative_slope=0.2)(x)
    x = Reshape((7, 7, 128))(x)

    x = Concatenate(axis=-1)([x, label_map])

    x = Conv2DTranspose(64, kernel_size=4, strides=2, padding="same", use_bias=False)(x)
    x = BatchNormalization()(x)
    x = LeakyReLU(negative_slope=0.2)(x)

    x = Conv2DTranspose(32, kernel_size=4, strides=2, padding="same", use_bias=False)(x)
    x = BatchNormalization()(x)
    x = LeakyReLU(negative_slope=0.2)(x)

    output_image = Conv2D(
        CHANNELS,
        kernel_size=3,
        strides=1,
        padding="same",
        activation="tanh"
    )(x)

    generator = Model([noise_input, label_input], output_image, name=name)

    return generator
````

Pada fungsi ini, generator menerima dua input, yaitu random noise dan label kelas. Random noise digunakan sebagai sumber variasi gambar, sedangkan label kelas digunakan agar generator menghasilkan gambar sesuai kelas tertentu, yaitu Parking Lot atau Storage Tanks.

Untuk setiap kelas, fungsi membuat sejumlah random noise dan label yang sama sesuai kelas yang sedang ditampilkan. Setelah itu, generator digunakan untuk menghasilkan fake image berdasarkan kombinasi noise dan label tersebut.

Output generator awalnya berada pada rentang -1 sampai 1 karena output layer menggunakan aktivasi `tanh`. Oleh karena itu, hasil gambar diubah kembali ke rentang 0 sampai 1 menggunakan rumus `(generated_images + 1) / 2` agar dapat divisualisasikan dengan benar menggunakan `imshow`.

Hasil gambar kemudian ditampilkan dalam bentuk grid menggunakan `matplotlib`. Dengan visualisasi ini, aku dapat melihat apakah modified generator mulai menghasilkan pola gambar yang menyerupai kelas Parking Lot dan Storage Tanks, atau masih menghasilkan gambar yang terlihat seperti noise.

### Build Modified Discriminator

````python
def build_modified_discriminator(
    num_classes=NUM_CLASSES,
    dropout_rate=0.3,
    name="Modified_Discriminator"
):
    image_input = Input(shape=(IMG_SIZE, IMG_SIZE, CHANNELS), name="image_input")
    label_input = Input(shape=(1,), dtype="int32", name="label_input")

    label_embedding = Embedding(num_classes, IMG_SIZE * IMG_SIZE)(label_input)
    label_embedding = Flatten()(label_embedding)
    label_map = Reshape((IMG_SIZE, IMG_SIZE, 1))(label_embedding)

    x = Concatenate(axis=-1)([image_input, label_map])

    x = Conv2D(64, kernel_size=4, strides=2, padding="same")(x)
    x = LeakyReLU(negative_slope=0.2)(x)
    x = Dropout(dropout_rate)(x)

    x = Conv2D(128, kernel_size=4, strides=2, padding="same")(x)
    x = LeakyReLU(negative_slope=0.2)(x)
    x = Dropout(dropout_rate)(x)

    x = Conv2D(256, kernel_size=3, strides=1, padding="same")(x)
    x = LeakyReLU(negative_slope=0.2)(x)
    x = Dropout(dropout_rate)(x)

    x = Flatten()(x)
    output = Dense(1, activation="sigmoid")(x)

    discriminator = Model([image_input, label_input], output, name=name)

    return discriminator
````

### Build Combined Modified CGAN

````python
def build_modified_cgan(generator, discriminator, noise_dim, generator_lr):
    discriminator.trainable = False

    noise_input = Input(shape=(noise_dim,), name="cgan_noise_input")
    label_input = Input(shape=(1,), dtype="int32", name="cgan_label_input")

    generated_image = generator([noise_input, label_input])
    validity = discriminator([generated_image, label_input])

    cgan = Model([noise_input, label_input], validity, name="Modified_Conditional_DCGAN")

    cgan.compile(
        loss="binary_crossentropy",
        optimizer=Adam(learning_rate=generator_lr, beta_1=0.5)
    )

    return cgan
````

### Penjelasan Build Modified Discriminator

Pada bagian ini, aku membangun **modified discriminator** untuk model Conditional DCGAN. Discriminator ini berfungsi untuk membedakan apakah gambar yang diberikan merupakan **real image** dari dataset atau **fake image** yang dihasilkan oleh generator. Karena model yang digunakan adalah Conditional GAN, discriminator tidak hanya menerima input gambar, tetapi juga menerima **label kelas** sebagai kondisi.

Discriminator menerima dua input utama, yaitu `image_input` dan `label_input`. `image_input` berisi gambar berukuran 28×28×1, sedangkan `label_input` berisi label kelas, yaitu Parking Lot atau Storage Tanks. Label ini digunakan agar discriminator dapat menilai gambar dengan mempertimbangkan kelas yang seharusnya diwakili oleh gambar tersebut.

Pada baseline discriminator sebelumnya, gambar diubah langsung menjadi vektor menggunakan `Flatten`, lalu diproses menggunakan Dense layer. Pada modified discriminator ini, aku menggunakan pendekatan yang berbeda, yaitu **Conv2D-based discriminator**. Tujuannya adalah agar discriminator dapat mempelajari pola spasial pada gambar, seperti garis, tekstur, edge, kontras objek, dan bentuk area tertentu. Pendekatan ini lebih cocok untuk data gambar dibandingkan hanya menggunakan Dense layer.

Label kelas diproses menggunakan `Embedding` layer dengan ukuran `IMG_SIZE * IMG_SIZE`. Karena gambar berukuran 28×28, maka embedding label dibuat menjadi 784 nilai. Setelah itu, hasil embedding diubah menggunakan `Reshape` menjadi bentuk 28×28×1. Bentuk ini disebut sebagai `label_map`.

Dengan cara ini, label tidak hanya menjadi vektor kecil, tetapi diubah menjadi peta spasial yang ukurannya sama dengan gambar. Setelah itu, `image_input` dan `label_map` digabungkan menggunakan `Concatenate(axis=-1)`. Karena keduanya memiliki ukuran 28×28×1, hasil penggabungannya menjadi 28×28×2. Artinya, discriminator menerima dua channel, yaitu satu channel gambar dan satu channel informasi label.

Setelah gambar dan label digabungkan, data masuk ke beberapa layer `Conv2D`. Layer pertama menggunakan 64 filter, kernel size 4, dan stride 2. Layer ini mulai menangkap pola visual awal pada gambar dan sekaligus mengecilkan ukuran feature map. Setelah itu digunakan aktivasi `LeakyReLU` dengan `negative_slope=0.2`. Aktivasi ini digunakan agar model dapat mempelajari pola non-linear dan tetap memiliki gradien kecil untuk nilai negatif.

Setelah aktivasi, digunakan `Dropout` dengan nilai sesuai parameter `dropout_rate`, yaitu 0.3. Dropout digunakan pada modified discriminator karena bagian 3C memang bertujuan melakukan modifikasi dan tuning. Tujuannya adalah mencegah discriminator terlalu cepat menjadi terlalu kuat atau terlalu dominan dibandingkan generator. Jika discriminator terlalu kuat, generator akan sulit belajar menghasilkan gambar yang lebih baik.

Layer convolution berikutnya menggunakan 128 filter dengan kernel size 4 dan stride 2. Layer ini mempelajari pola visual yang lebih kompleks dibanding layer pertama. Setelah itu kembali digunakan `LeakyReLU` dan `Dropout`.

Layer convolution ketiga menggunakan 256 filter dengan kernel size 3 dan stride 1. Layer ini tidak lagi mengecilkan ukuran feature map secara besar, tetapi lebih berfokus pada memperkaya representasi fitur. Dengan jumlah filter yang lebih besar, discriminator memiliki kapasitas lebih tinggi untuk mengenali perbedaan antara real image dan fake image.

Setelah melewati convolutional layer, hasil feature map diubah menjadi vektor menggunakan `Flatten`. Vektor ini kemudian masuk ke output layer berupa `Dense(1, activation="sigmoid")`. Output sigmoid menghasilkan satu nilai probabilitas. Nilai mendekati 1 berarti discriminator menganggap gambar tersebut real, sedangkan nilai mendekati 0 berarti gambar dianggap fake.

Secara keseluruhan, modified discriminator ini dibuat untuk meningkatkan kemampuan model dalam mengenali pola gambar secara spasial. Perbedaan utama dibanding baseline adalah penggunaan `Conv2D`, label map, dan Dropout. Dengan arsitektur ini, discriminator diharapkan dapat memberikan feedback yang lebih baik kepada generator, sehingga generator dapat belajar menghasilkan fake image yang distribusinya lebih mendekati real image.

### Helper untuk Loss

````python
def to_scalar(value):
    if isinstance(value, (list, tuple)):
        return float(value[0])
    return float(value)
````

### Custom Training Function Modified DCGAN

````python
def train_modified_cgan(
    generator,
    discriminator,
    cgan,
    X_train,
    y_train,
    config,
    display_interval=25
):
    noise_dim = config["noise_dim"]
    epochs = config["epochs"]
    batch_size = config["batch_size"]
    steps_per_epoch = config["steps_per_epoch"]
    real_label_value = config["real_label_value"]

    half_batch = batch_size // 2

    d_losses = []
    d_accuracies = []
    g_losses = []

    for epoch in range(1, epochs + 1):
        epoch_d_loss = []
        epoch_d_acc = []
        epoch_g_loss = []

        for step in range(steps_per_epoch):
            # Train discriminator with real images
            real_idx = np.random.randint(0, X_train.shape[0], half_batch)
            real_images = X_train[real_idx]
            real_labels = y_train[real_idx].reshape(-1, 1)
            real_targets = np.full((half_batch, 1), real_label_value)

            # Train discriminator with fake images
            noise = np.random.normal(0, 1, (half_batch, noise_dim)).astype("float32")
            fake_labels = np.random.randint(0, NUM_CLASSES, half_batch).reshape(-1, 1)
            fake_images = generator.predict([noise, fake_labels], verbose=0)
            fake_targets = np.zeros((half_batch, 1))

            discriminator.trainable = True

            d_loss_real = discriminator.train_on_batch(
                [real_images, real_labels],
                real_targets
            )

            d_loss_fake = discriminator.train_on_batch(
                [fake_images, fake_labels],
                fake_targets
            )

            d_loss = 0.5 * np.add(d_loss_real, d_loss_fake)

            # Train generator through combined CGAN
            noise = np.random.normal(0, 1, (batch_size, noise_dim)).astype("float32")
            sampled_labels = np.random.randint(0, NUM_CLASSES, batch_size).reshape(-1, 1)
            valid_targets = np.ones((batch_size, 1))

            discriminator.trainable = False

            g_loss = cgan.train_on_batch(
                [noise, sampled_labels],
                valid_targets
            )

            epoch_d_loss.append(float(d_loss[0]))
            epoch_d_acc.append(float(d_loss[1]))
            epoch_g_loss.append(to_scalar(g_loss))

        d_losses.append(np.mean(epoch_d_loss))
        d_accuracies.append(np.mean(epoch_d_acc))
        g_losses.append(np.mean(epoch_g_loss))

        if epoch % display_interval == 0 or epoch == 1:
            print(
                f"{config['name']} | "
                f"Epoch {epoch}/{epochs} | "
                f"D Loss: {d_losses[-1]:.4f} | "
                f"D Acc: {d_accuracies[-1]:.4f} | "
                f"G Loss: {g_losses[-1]:.4f}"
            )

            show_modified_generated_images(
                generator=generator,
                epoch=epoch,
                noise_dim=noise_dim,
                title_prefix=config["name"]
            )

    history = {
        "d_losses": d_losses,
        "d_accuracies": d_accuracies,
        "g_losses": g_losses
    }

    return history
````

### Penjelasan Training Function Modified CGAN

Fungsi `train_modified_cgan()` digunakan untuk melatih modified Conditional GAN secara manual. Training GAN tidak menggunakan `model.fit()` seperti model deep learning biasa, karena generator dan discriminator harus dilatih secara bergantian. Oleh karena itu, fungsi ini dibuat untuk mengatur proses training discriminator dan generator dalam setiap epoch.

Fungsi ini menerima beberapa input utama, yaitu `generator`, `discriminator`, `cgan`, data training `X_train` dan `y_train`, serta `config` yang berisi hyperparameter seperti `noise_dim`, `epochs`, `batch_size`, `steps_per_epoch`, dan `real_label_value`.

Pada awal fungsi, nilai hyperparameter diambil dari `config`. `noise_dim` digunakan untuk menentukan ukuran random noise yang masuk ke generator. `epochs` menentukan jumlah perulangan training. `batch_size` menentukan jumlah data yang digunakan dalam satu batch. `steps_per_epoch` menentukan berapa kali update dilakukan dalam satu epoch. Sementara itu, `real_label_value` digunakan untuk one-sided label smoothing, yaitu memberi label real dengan nilai 0.9, bukan 1.0, agar discriminator tidak terlalu percaya diri.

Variabel `half_batch` digunakan karena pada setiap step, discriminator dilatih menggunakan dua jenis data, yaitu real image dan fake image. Setengah batch digunakan untuk real image, dan setengah batch lainnya digunakan untuk fake image. Dengan cara ini, discriminator belajar membedakan gambar asli dari dataset dan gambar palsu dari generator.

Pada setiap epoch, fungsi menyiapkan list sementara untuk menyimpan nilai loss discriminator, accuracy discriminator, dan loss generator. Nilai-nilai ini kemudian dirata-ratakan pada akhir epoch agar dapat digunakan untuk memantau perkembangan training.

Pada bagian pertama setiap step, discriminator dilatih menggunakan real image. Real image diambil secara acak dari `X_train`, sedangkan label kelasnya diambil dari `y_train`. Target untuk real image dibuat menggunakan `real_label_value`, misalnya 0.9. Ini berarti discriminator dilatih untuk mengenali gambar asli sebagai real, tetapi dengan label yang sedikit lebih lembut agar training lebih stabil.

Pada bagian berikutnya, generator membuat fake image. Random noise dibuat menggunakan distribusi normal, lalu fake label dibuat secara acak dari jumlah kelas yang tersedia, yaitu Parking Lot dan Storage Tanks. Noise dan fake label tersebut diberikan ke generator untuk menghasilkan fake image. Target untuk fake image dibuat bernilai 0, karena discriminator harus mengenali fake image sebagai gambar palsu.

Setelah real image dan fake image disiapkan, discriminator dibuat `trainable = True`. Hal ini dilakukan karena pada tahap ini bobot discriminator memang ingin diperbarui. Discriminator kemudian dilatih dua kali, yaitu sekali menggunakan real image dan sekali menggunakan fake image. Loss dari real image dan fake image kemudian dirata-ratakan agar diperoleh total discriminator loss pada step tersebut.

Setelah discriminator dilatih, tahap berikutnya adalah melatih generator melalui combined CGAN. Pada tahap ini, random noise dan label baru dibuat lagi sebanyak `batch_size`. Target yang digunakan adalah 1, karena tujuan generator adalah menghasilkan fake image yang dapat dianggap real oleh discriminator.

Sebelum melatih generator, discriminator dibuat `trainable = False`. Hal ini penting karena saat training combined CGAN, yang ingin diperbarui hanya bobot generator. Discriminator hanya digunakan sebagai pemberi feedback, bukan sebagai model yang ikut dilatih. Dengan cara ini, generator belajar memperbaiki kualitas fake image berdasarkan penilaian discriminator.

Nilai loss dari discriminator dan generator disimpan pada setiap step. Setelah semua step dalam satu epoch selesai, nilai loss dan accuracy dirata-ratakan. Rata-rata tersebut kemudian disimpan ke dalam list `d_losses`, `d_accuracies`, dan `g_losses`.

Setiap beberapa epoch sesuai `display_interval`, fungsi akan menampilkan informasi training seperti discriminator loss, discriminator accuracy, dan generator loss. Selain itu, fungsi `show_modified_generated_images()` juga dipanggil untuk menampilkan contoh fake image yang dihasilkan generator. Visualisasi ini penting karena evaluasi GAN tidak cukup hanya dari loss, tetapi juga perlu melihat kualitas gambar yang dihasilkan.

Pada akhir training, fungsi mengembalikan dictionary `history` yang berisi riwayat discriminator loss, discriminator accuracy, dan generator loss. Riwayat ini dapat digunakan untuk membuat grafik training dan menganalisis apakah proses training berjalan stabil atau tidak.

Secara keseluruhan, fungsi ini mengatur proses training modified CGAN secara manual dengan urutan: melatih discriminator menggunakan real dan fake image, membekukan discriminator, lalu melatih generator melalui combined CGAN. Proses ini sesuai dengan cara kerja GAN, yaitu generator dan discriminator belajar secara bergantian dalam proses adversarial training.

### Generate Fake Images by Labels

````python
def generate_images_by_labels(
    generator,
    labels,
    noise_dim,
    batch_size=256,
    seed=SEED
):
    rng = np.random.default_rng(seed)
    generated_images = []

    labels = labels.reshape(-1, 1)

    for start in range(0, len(labels), batch_size):
        end = start + batch_size
        current_labels = labels[start:end]

        noise = rng.normal(0, 1, (len(current_labels), noise_dim)).astype("float32")
        fake_batch = generator.predict([noise, current_labels], verbose=0)

        generated_images.append(fake_batch)

    generated_images = np.concatenate(generated_images, axis=0)

    return generated_images
````

### FID Function

````python
def prepare_images_for_inception(images):
    images = ((images + 1) * 127.5).astype("float32")
    images = np.clip(images, 0, 255)

    images = tf.convert_to_tensor(images)
    images = tf.image.grayscale_to_rgb(images)
    images = tf.image.resize(images, (75, 75))
    images = preprocess_input(images)

    return images.numpy()


def get_inception_features(inception_model, images, batch_size=128):
    prepared_images = prepare_images_for_inception(images)

    features = inception_model.predict(
        prepared_images,
        batch_size=batch_size,
        verbose=1
    )

    return features


def calculate_fid_from_features(real_features, fake_features, eps=1e-6):
    mu_real = np.mean(real_features, axis=0)
    mu_fake = np.mean(fake_features, axis=0)

    sigma_real = np.cov(real_features, rowvar=False)
    sigma_fake = np.cov(fake_features, rowvar=False)

    sigma_real += np.eye(sigma_real.shape[0]) * eps
    sigma_fake += np.eye(sigma_fake.shape[0]) * eps

    mean_diff = mu_real - mu_fake
    covmean = sqrtm(sigma_real.dot(sigma_fake))

    if np.iscomplexobj(covmean):
        covmean = covmean.real

    fid = mean_diff.dot(mean_diff) + np.trace(
        sigma_real + sigma_fake - 2 * covmean
    )

    return float(fid)
````

Parameter `pooling="avg"` digunakan agar output feature map diringkas menjadi satu vektor fitur. Fitur real image ini disimpan sebagai `real_test_features`. Nantinya fitur ini akan dibandingkan dengan fitur fake image dari generator untuk menghitung **Fréchet Inception Distance (FID)**. Semakin kecil nilai FID, semakin mirip distribusi fake image dengan distribusi real image.

Dataset yang digunakan berisi gambar 28×28 grayscale dengan dua kelas, yaitu Parking Lot dan Storage Tanks. Karena ukuran gambar relatif kecil, hyperparameter tidak dibuat terlalu besar agar model tetap efisien. Namun, karena baseline MLP CGAN masih menghasilkan gambar yang kurang jelas, aku mencoba beberapa variasi pada modified Conditional DCGAN untuk melihat konfigurasi mana yang dapat menghasilkan distribusi fake image paling mendekati real image berdasarkan nilai FID.

#### Isi Hyperparameter pada Setiap Config

Setiap konfigurasi memiliki beberapa parameter penting:

`name` digunakan sebagai nama eksperimen. Nama ini dibuat berbeda untuk setiap konfigurasi agar hasil training, model yang disimpan, dan nilai FID dapat dibedakan dengan mudah.

`noise_dim` adalah ukuran random noise yang menjadi input generator. Noise digunakan sebagai sumber variasi agar generator tidak menghasilkan gambar yang sama terus-menerus. Pada eksperimen ini, aku mencoba `noise_dim` 100 dan 128. Nilai 100 digunakan sebagai ukuran umum pada GAN, sedangkan 128 digunakan untuk memberi ruang latent yang sedikit lebih besar kepada generator.

`batch_size` adalah jumlah data yang digunakan dalam satu proses update. Batch size yang lebih besar seperti 128 dapat membuat training lebih stabil karena update bobot dihitung dari lebih banyak sampel. Sementara itu, batch size yang lebih kecil seperti 64 membuat update lebih sering dan bisa membuat model lebih adaptif, tetapi training dapat menjadi lebih tidak stabil dan lebih lama.

`steps_per_epoch` menunjukkan berapa kali proses update dilakukan dalam satu epoch. Untuk batch size 128, aku menggunakan 100 steps per epoch agar training tetap efisien. Untuk batch size 64, aku menggunakan 200 steps per epoch karena ukuran batch lebih kecil, sehingga jumlah step perlu ditambah agar jumlah data yang dipelajari dalam satu epoch tetap relatif sebanding.

`epochs` adalah jumlah total perulangan training. Pada semua eksperimen modified, aku menggunakan 150 epoch. Jumlah ini dipilih karena GAN membutuhkan waktu yang cukup panjang untuk belajar menghasilkan gambar, tetapi tidak dibuat terlalu besar agar training tetap realistis dan tidak terlalu memakan waktu.

`generator_lr` adalah learning rate untuk generator. Nilai ini menentukan seberapa besar perubahan bobot generator pada setiap update. Jika learning rate terlalu besar, generator bisa menghasilkan gambar yang tidak stabil. Jika terlalu kecil, generator bisa belajar terlalu lambat. Pada eksperimen ini, aku mencoba nilai 0.0002 dan 0.0001.

`discriminator_lr` adalah learning rate untuk discriminator. Pada modified model ini, discriminator learning rate dibuat 0.0001. Nilai ini sengaja dibuat lebih kecil atau sama dengan generator learning rate agar discriminator tidak terlalu cepat menjadi terlalu kuat. Jika discriminator terlalu dominan, generator akan sulit belajar karena fake image terlalu mudah dikenali sebagai fake.

`real_label_value` digunakan untuk one-sided label smoothing. Pada training discriminator, real image tidak diberi label 1.0, tetapi diberi label 0.9. Tujuannya adalah agar discriminator tidak terlalu percaya diri saat membedakan real image dan fake image. Teknik ini dapat membantu training GAN menjadi lebih stabil.

`dropout_rate` adalah nilai dropout pada modified discriminator. Pada eksperimen ini, dropout rate dibuat 0.3. Dropout digunakan untuk mencegah discriminator terlalu cepat overfit atau terlalu dominan. Hal ini penting karena dalam GAN, generator membutuhkan feedback yang seimbang dari discriminator agar dapat belajar menghasilkan gambar yang lebih baik.

### Train Semua Modified Experiments

````python
modified_results = []
modified_generators = {}
modified_discriminators = {}
modified_histories = {}

for config in MODIFIED_CONFIGS:
    print("=" * 80)
    print("Training:", config["name"])
    print("=" * 80)

    generator = build_modified_generator(
        noise_dim=config["noise_dim"],
        name=f"{config['name']}_Generator"
    )

    discriminator = build_modified_discriminator(
        dropout_rate=config["dropout_rate"],
        name=f"{config['name']}_Discriminator"
    )

    discriminator.compile(
        loss="binary_crossentropy",
        optimizer=Adam(learning_rate=config["discriminator_lr"], beta_1=0.5),
        metrics=["accuracy"]
    )

    cgan = build_modified_cgan(
        generator=generator,
        discriminator=discriminator,
        noise_dim=config["noise_dim"],
        generator_lr=config["generator_lr"]
    )

    print("\nGenerator Summary:")
    generator.summary()

    print("\nDiscriminator Summary:")
    discriminator.summary()

    history = train_modified_cgan(
        generator=generator,
        discriminator=discriminator,
        cgan=cgan,
        X_train=X_train,
        y_train=y_train,
        config=config,
        display_interval=25
    )

    plot_training_history(history, config["name"])

    X_fake_test_modified = generate_images_by_labels(
        generator=generator,
        labels=y_test,
        noise_dim=config["noise_dim"],
        batch_size=256,
        seed=SEED
    )

    fake_test_features = get_inception_features(
        inception_model=inception_model,
        images=X_fake_test_modified,
        batch_size=128
    )

    fid_score = calculate_fid_from_features(
        real_features=real_test_features,
        fake_features=fake_test_features
    )

    print(f"{config['name']} FID on test data: {fid_score:.4f}")

    generator_path = MODEL_DIR / f"{config['name']}_generator.keras"
    discriminator_path = MODEL_DIR / f"{config['name']}_discriminator.keras"

    generator.save(str(generator_path))
    discriminator.save(str(discriminator_path))

    modified_results.append({
        "name": config["name"],
        "noise_dim": config["noise_dim"],
        "batch_size": config["batch_size"],
        "steps_per_epoch": config["steps_per_epoch"],
        "epochs": config["epochs"],
        "generator_lr": config["generator_lr"],
        "discriminator_lr": config["discriminator_lr"],
        "real_label_value": config["real_label_value"],
        "dropout_rate": config["dropout_rate"],
        "fid": fid_score,
        "generator_path": str(generator_path),
        "discriminator_path": str(discriminator_path)
    })

    modified_generators[config["name"]] = generator
    modified_discriminators[config["name"]] = discriminator
    modified_histories[config["name"]] = history
````

### Alasan Modified Model Lebih Baik daripada Baseline

Modified model lebih baik daripada baseline karena arsitektur yang digunakan lebih sesuai dengan karakter data gambar. Pada baseline, model masih menggunakan **MLP CGAN** yang berbasis Dense layer. Dense layer memperlakukan gambar sebagai deretan nilai piksel dalam bentuk vektor, sehingga hubungan spasial antar piksel tidak dipertahankan secara kuat. Akibatnya, model baseline kurang mampu menangkap pola visual seperti garis, tekstur, edge, area kontras, dan bentuk objek pada gambar Overhead-MNIST.

Sebaliknya, modified model menggunakan **Conditional DCGAN** yang berbasis convolutional layer. Pada generator, digunakan `Conv2DTranspose` untuk membangun gambar secara bertahap dari feature map kecil menjadi gambar berukuran 28×28. Proses ini lebih natural untuk image generation karena generator tidak langsung menghasilkan semua piksel melalui Dense layer, tetapi belajar membentuk struktur gambar secara bertahap. Hal ini membuat fake image yang dihasilkan modified model lebih berpotensi memiliki pola visual yang lebih terstruktur dibandingkan baseline.

Pada discriminator, modified model menggunakan `Conv2D`. Layer convolution lebih cocok untuk data gambar karena dapat menangkap pola lokal pada citra, seperti garis parkir pada kelas Parking Lot, tekstur area parkir, edge, serta perbedaan area terang dan gelap pada Storage Tanks. Dengan kemampuan ini, discriminator dapat memberikan feedback yang lebih baik kepada generator. Feedback yang lebih baik membantu generator belajar menghasilkan fake image yang distribusinya lebih mendekati real image.

Modified model juga tetap mempertahankan konsep **Conditional GAN**, yaitu menggunakan label kelas pada generator dan discriminator. Hal ini penting karena dataset hanya terdiri dari dua kelas, yaitu Parking Lot dan Storage Tanks. Dengan class label, generator dapat diarahkan untuk menghasilkan gambar berdasarkan kelas tertentu, sedangkan discriminator dapat menilai real/fake image dengan mempertimbangkan label kelas yang diberikan.

Selain perubahan arsitektur, modified model juga menggunakan teknik tambahan yang membantu training lebih stabil. **Batch Normalization** digunakan pada generator untuk menjaga distribusi aktivasi agar proses training lebih stabil. **LeakyReLU** digunakan agar model tetap memiliki gradien pada nilai negatif. **Dropout** digunakan pada discriminator agar discriminator tidak terlalu cepat menjadi terlalu dominan dibandingkan generator. Selain itu, **one-sided label smoothing** dengan real label 0.9 digunakan agar discriminator tidak terlalu percaya diri dalam membedakan real image dan fake image.

Hasil evaluasi menunjukkan bahwa modified model memang lebih baik daripada baseline. Baseline MLP CGAN menghasilkan nilai FID sebesar **1159.4017**, sedangkan best modified model, yaitu **Modified\_DCGAN\_Exp2\_Noise128\_BS128**, menghasilkan nilai FID sebesar **1111.724735**. Karena nilai FID yang lebih rendah menunjukkan distribusi fake image yang lebih mendekati distribusi real image, maka penurunan FID ini menunjukkan bahwa modified model berhasil memperbaiki kualitas distribusi gambar yang dihasilkan.

Peningkatan performa ini juga didukung oleh hasil tuning hyperparameter. Pada best modified model, noise dimension dinaikkan menjadi 128 sehingga generator memiliki ruang latent yang lebih luas untuk menghasilkan variasi gambar. Batch size 128 membantu proses training tetap stabil, sedangkan kombinasi generator learning rate 0.0002 dan discriminator learning rate 0.0001 menjaga agar generator tetap dapat belajar tanpa terlalu cepat dikalahkan oleh discriminator.

Dengan demikian, modified model lebih baik daripada baseline karena tiga alasan utama. Pertama, arsitektur Conditional DCGAN lebih sesuai untuk data gambar dibandingkan MLP CGAN. Kedua, penggunaan convolutional layer membuat model lebih mampu menangkap struktur spasial pada citra. Ketiga, hasil evaluasi FID membuktikan bahwa distribusi fake image dari modified model lebih mendekati distribusi real image dibandingkan baseline.
